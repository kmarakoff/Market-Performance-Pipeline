```python
import pandas as pd
url = 'https://docs.google.com/spreadsheets/d/e/2PACX-1vSBwifjcwZpKjVyQQrIxtHk0WFmgyGUsY4ou60F4m4849vUNPOJtV8ohCjKSYqXCr83UXnD5a1_StwY/pub?gid=0&single=true&output=csv'

raw_data = pd.read_csv(url)
df = raw_data
df.head()
```

```python
df.columns
```

```python
df['Wins'] = pd.to_numeric(df['Wins'], errors='coerce')
df['Losses'] = pd.to_numeric(df['Losses'], errors='coerce')
df['Goals For (GF)'] = pd.to_numeric(df['Goals For (GF)'], errors='coerce')
```

```python
df['Wins'] = df.groupby('Year')['Wins'].transform(lambda x: x.fillna(x.median()))
df['Losses'] = df.groupby('Year')['Losses'].transform(lambda x: x.fillna(x.median()))
df['Goals For (GF)'] = df.groupby('Year')['Goals For (GF)'].transform(lambda x: x.fillna(x.median()))
```

```python
import datetime
df['scraped_date'] = datetime.datetime.now().strftime('%Y-%m-%d')
final_dataset = df[['Team Name', 'Year', 'Wins', 'Losses', 'Goals For (GF)', 'scraped_date']]
final_dataset.to_csv('clean_performance_data.csv', index=False)

print("Export Success!")
```
